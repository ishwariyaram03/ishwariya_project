```python
# Replace with item-count columns
cols = [
   'DAYS_FIRST_DRAWING',
   'DAYS_FIRST_DUE',
   'DAYS_LAST_DUE_1ST_VERSION',
   'DAYS_LAST_DUE',
   'DAYS_TERMINATION'
]

group_col = 'NAME_CONTRACT_STATUS'

for col in cols:
  if col in df.columns:
    # 1. Flag rows that had missing values before imputing
    df[f'{col}_missing_flag'] = df[col].isna().astype(int)

    # 2. Compute group-wise mode aligned to df's index
    group_mode = df.groupby(group_col)[col].transform(
        lambda x: (
            x.mode(dropna=True)[0]
            if not x.mode(dropna=True).empty
            else (df[col].mode(dropna=True)[0] if not df[col].mode(dropna=True).empty else np.nan)
        )
    )

    # 3. Fill missing values and verify
    df[col] = df[col].fillna(group_mode)
    print(f"{col}: remaining nulls = {df[col].isna().sum()}, flagged = {df[f'{col}_missing_flag'].sum()}")
```

```python
# Hypothesis definition (written before running the test, which is best practice)
# Define H0 and H1
# H0: There is no significant difference in the mean delivery charge across different traffic density levels.
# H1: There is a significant difference in the mean delivery charge across different traffic density levels
# Alpha=0.05
# if p-value<0.05, Reject H0, Accept H1-->THere is a significant difference in the mean delivery charge across different traffic density levels.
# else, Accept H0, Reject H1-->There is no significant difference in the mean delivery charge across different traffic density levels.
```

```python
# Group AMT_CREDIT by each unique NAME_GOODS_CATEGORY value
class_cat = []
for cls in df['NAME_GOODS_CATEGORY'].unique():
    group_data = df[df['NAME_GOODS_CATEGORY'] == cls]['AMT_CREDIT']
    class_cat.append(group_data)

# Run one-way ANOVA
from scipy.stats import f_oneway

f_stat, p_val = f_oneway(*class_cat)
alpha = 0.05

print(f"F-statistic: {f_stat}, P-value: {p_val}")
if p_val < alpha:
    print("Reject H0: There is a significant difference in the AMT_CREDIT across GOODSCATEGORY.")
else:
    print("Accept H0: There is no significant difference in the AMT_CREDIT across GOODSCATEGORY.")
```

```python
import pandas as pd
from scipy.stats import ttest_ind, f_oneway

# Reusable functions: run the test and return a result dictionary
# Selection_Status = 'YES' if p < alpha (feature significantly affects the target), otherwise 'NO'
# ANOVA_Test -> features with 3+ classes; Two_Sample_T_Test -> features with exactly 2 classes
def ANOVA_Test(df, Feat_clm, Target_clm, alpha=0.05):
    class_cat=[]
    for cls in df[Feat_clm].unique():
        class_cat.append(df[df[Feat_clm]==cls][Target_clm])

    f_stat, p_val = f_oneway(*class_cat) # Perform ANOVA test

    if p_val < alpha:
       return {'Feature_Column': Feat_clm,
               'Target_Column': Target_clm,
                'Test-statistic': f_stat,
                'P-value': p_val,
                'Selection_Status':'YES',
                'Test_Type':'ANOVA Test'}
    else:
           return {'Feature_Column': Feat_clm,
               'Target_Column': Target_clm,
                'Test-statistic': f_stat,
                'P-value': p_val,
                'Selection_Status':'NO',
                'Test_Type':'ANOVA Test'}


def Two_Sample_T_Test(df, Feat_clm, Target_clm, alpha=0.05):
    unique_groups = df[Feat_clm].unique()
    if len(unique_groups) > 2:
        # Automatically hand over to ANOVA if there are more than 2 categories
        return ANOVA_Test(df, Feat_clm, Target_clm, alpha)

    class_cat=[]
    for cls in unique_groups:
        class_cat.append(df[df[Feat_clm]==cls][Target_clm])

    t_stat, p_val = ttest_ind(*class_cat) # Perform 2-Sample T-test

    if p_val < alpha:
       return {'Feature_Column': Feat_clm,
               'Target_Column': Target_clm,
                'Test-statistic': t_stat,
                'P-value': p_val,
                'Selection_Status':'YES',
                'Test_Type':'Two-Sample T-Test'}
    else:
           return {'Feature_Column': Feat_clm,
               'Target_Column': Target_clm,
                'Test-statistic': t_stat,
                'P-value': p_val,
                'Selection_Status':'NO',
                'Test_Type':'Two-Sample T-Test'}

Two_Sample_T_Test(df, 'NAME_GOODS_CATEGORY', 'AMT_CREDIT')
```

```python
# Automatic test selection: 2 classes -> T-test, 3+ classes -> ANOVA
Cat_Selected_list=[]
for col in Cat_columns:
    # Skip columns that have fewer than 2 categories because statistical tests require at least 2 groups
    n_unique = df[col].nunique()
    if n_unique < 2:
        continue

    if n_unique == 2:
        result = Two_Sample_T_Test(df, col, 'AMT_CREDIT')
    else:
        result = ANOVA_Test(df, col, 'AMT_CREDIT')
    Cat_Selected_list.append(result)

Cat_Selected_list
```

```python
# Results as a table: test statistic, p-value and selection decision per feature
Cat_Selected_df=pd.DataFrame(Cat_Selected_list)
Cat_Selected_df
```

```python
#Categorical VS cateogorical #chi square Test
from scipy.stats import chi2_contingency

results = []
target_columns = ['NAME_CONTRACT_STATUS', 'CODE_REJECT_REASON', 'NAME_CLIENT_TYPE']

for target_col in target_columns:
  for clmn in Cat_columns:
    pc = pd.crosstab(df[clmn], df[target_col])
    chi2, p, dof, expected = chi2_contingency(pc)
    selected = 'yes' if p < 0.05 else 'no'
    results.append({
        'target': target_col,
        'clmn': clmn,
        'chi2': chi2,
        'p': p,
        'selected': selected,
    })

    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x=clmn, hue=target_col)
    plt.title(f'{target_col} by {clmn}')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    #plt.savefig(f'countplot_{target_col}_{clmn}.png', dpi=120)
    plt.show()
```
